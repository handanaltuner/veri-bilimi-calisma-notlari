```python
#Gizli sayıyı belirleme
X=7
#Ekrana mesaj yazdırıp kullanıcıdan sayı isteriz (Y)
print("Aklımdan bir sayı tuttum, hadi onu tahmin et :)")
Y_input=input("Lütfen bir sayı gir: ")
```

```python
#Kullanıcınn girdiği metni tam sayıya çevirme
Y=int(Y_input)
```

```python
#Sayıyı kendisiyle çarpıp, kendisiyle çıkarıyoruz.
sonuc=(Y*Y-Y)
#Çıkan sonuc ile gizli sayı arasındaki farkı hesaplama
fark=abs(sonuc-X)
#Kosulları kontrol etme
if sonuc ==X:
  print("Sonuç:", sonuc, "Tebrikler, 100 puan kazandın.")
elif fark <=20:
  print("Sonuç:", sonuc, "Çok yaklaştın, 50 puan kazandın.")
elif fark<=50:
  print("Sonuç:", sonuc, "Yaklaştın, yeniden dene :()")
else:
  print("Sonuç:", sonuc, "Kaybettin :()")
```
