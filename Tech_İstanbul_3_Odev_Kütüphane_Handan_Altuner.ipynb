```python
DOSYA_ADI="kutuphane.txt"

def dosyadan_oku():
  kitaplar=[]
  try:
    with open(DOSYA_ADI, "r", encoding="utf-8") as dosya:
      for satir in dosya:
        satir = satir.strip()
        if not satir:
          continue
        parcalar=satir.strip().split(";")
        if len(parcalar)==2:
          kitaplar.append({"Kitap_adi": parcalar[0], "yazar": parcalar[1]})
  except FileNotFoundError:
    print(f"Dosya bulunamadı: {DOSYA_ADI}")
  return kitaplar
```

```python
 # Kütüphane Uygulaması
# Kütüphanede baştan bulunan kitaplar (her kitap bir sözlük: ad ve yazar)
kitaplar=dosyadan_oku()
```

```python
#Kitapları listele
def kitaplari_listele():
  if len (kitaplar)==0:
    print("Kütüphanede kayıtlı kitap bulunmamaktadır.")
    return


  for i, kitap in enumerate(kitaplar, start=1):
      print(f"{i}-{kitap['Kitap_adi']} - {kitap['yazar']}")
```

```python
#Kitap ekle
def kitap_ekle():
  kitap_adi =input("Kitabın adı: ")
  yazar =input("Yazarın adı: ")

  #Aynı kitap kütüphanede kayıtlı olarak var mı? Onu kontrol et
  for kitap in kitaplar:
    if kitap["Kitap_adi"].lower() ==  kitap_adi.lower() and kitap["yazar"].lower()==yazar.lower():
      print("Bu kitap zaten kütüphanede kayıtlı.")
      return

  kitaplar.append({"Kitap_adi":kitap_adi, "yazar":yazar})


  with open(DOSYA_ADI,"a", encoding="utf-8") as dosya:

    dosya.write(f"{kitap_adi};{yazar}\n")

  print("Kitaplar kütüphaneye başarı ile eklendi")

  kitaplari_listele()


```

```python
def dosyaya_yeniden_yaz():
  with open(DOSYA_ADI, "w", encoding="utf-8") as f:
    for kitap in kitaplar:
      f.write(f"{kitap['Kitap_adi']}; {kitap['yazar']}\n")
```

```python
#Kitap çıkar
def kitap_cikar():
  aranan_yazar=input("Almak istediğiniz kitabın yazarını girin: ")
  aranan_kitap=input("Almak istediğiniz kitabın adını girin: ")

  #Bu yazara ait kitapları bul
  yazarin_kitaplari=[
    kitap for kitap in kitaplar
    if kitap["yazar"].lower()==aranan_yazar.lower()
  ]

  if len (yazarin_kitaplari)==0:
    print("Bu yazarın kitapları kütüphanede bulunmamaktadır.")
    return

#Bulunan kitapları göster
  for i, kitap in enumerate(yazarin_kitaplari, start=1):
    print(f"{i}-{kitap['Kitap_adi']}")

 #aranan kitapları bul
  aranan_kitaplar=[
  kitap for kitap in yazarin_kitaplari
  if kitap["Kitap_adi"].lower()== aranan_kitap.lower()

 ]

  if len (aranan_kitaplar)==0:
    print("Bu kitap kütüphanede bulunmamaktadır.")
    return

  #numaralı listeyi göster
  for i, kitap in enumerate(aranan_kitaplar, start=1):
     print(f"{i}-{kitap['Kitap_adi']}")

  secim=input("Almak istediğiniz kitabın numarasını girin: ")

  #Secim gecerliliği kontrolu

  if not secim.isdigit() or int(secim) < 1 or int(secim) > len(aranan_kitaplar):
    print("Geçersiz seçim.")
    return

#Kitabı listeden cıkar ve bilgi ve
  secilen_kitap=aranan_kitaplar[int(secim)-1]
  kitaplar.remove(secilen_kitap)
  dosyaya_yeniden_yaz()

  print(f"{secilen_kitap['Kitap_adi']} adli kitabı aldınız. İyi okumalar")





```

```python
def menu_goster():
  print("\n### Kütüphane Yönetim Menüsü ###")
  print("1. Kitap Ekle")
  print("2. Kitap Çıkar")
  print("3. Kitapları Listele")
  print("4. Çıkış")

#Ana program döngüsü
while True:
    menu_goster()
    secim = input("Seçiminizi girin: ")

    if secim == "1":
        kitap_ekle()
    elif secim == "2":
        kitap_cikar()
    elif secim == "3":
        kitaplari_listele()
    elif secim == "4":
        print("Programdan çıkılıyor.")
        break
    else:
        print("Geçersiz seçim, lütfen 1-4 arası bir sayı girin.")
```
